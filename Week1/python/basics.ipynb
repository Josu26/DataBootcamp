```python
import pandas as pd
import matplotlib.pyplot as plt

print("Todo listo, Josué 🚀")

df = pd.DataFrame({
    'product': ['A','B','C'],
    'revenue': [1200, 800, 600],
    'cost':    [500, 300, 200]
})
df['profit'] = df['revenue'] - df['cost']
df
```

```python
ax = df.plot(kind='bar', x='product', y='profit', color='skyblue', legend=False)
ax.set_title('Profit por producto')
ax.set_xlabel('Producto')
ax.set_ylabel('Profit (€)')
plt.show()
```
